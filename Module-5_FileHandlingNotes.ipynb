#READING TEXT FILES

Reading a text file means retrieving the data stored in a file without changing it.

Python uses open() in read mode ('r').

# Methods:

read() - Reads entire file

readline() - Reads one line

readlines() - Reads all lines as a list

# WRITING TEXT FILES

Writing stores data into a text file.

Modes:

w - Write (overwrite)

a - Append(add at the end)

x - Create new file

```python
sampletxt = open("sample.txt", "w")
sampletxt.write("Hello World.\n")
sampletxt.write("This is Infant.\n")
sampletxt.write("This file is my Sprint 1 project.")
sampletxt.close()

print("File created")
```

```python
sampletxt = open("sample.txt", "r")
data = sampletxt.read()
print(data)
sampletxt.close()
```

# CSV FILES
CSV (Comma-Separated Values) stores tabular data. Each value is separated by a comma.

Uses: Student records, employee data, sales reports.

# Example:

Name, Age, Marks

Infy, 33, 97

Deepan, 34, 91

```python
import pandas as pd

data = {
    "Name": ["Infy", "Deepan", "Prisci"],
    "Age": [33, 34, 29],
    "City": ["Chennai", "Dharamapuri", "Bangalore"]
}

df = pd.DataFrame(data)

df.to_csv("students.csv", index=False)

print("CSV file created.")
```

```python
import pandas as pd

df = pd.read_csv("students.csv")

print(df)
```

# JSON FILES

JSON (JavaScript Object Notation) stores data in key-value pairs.

Widely used in APIs and configuration files.

Example:

{ "Name":"Infy" "Age" : 33 }

```python
import json

data = {
    "name": "Infy",
    "age": 33,
    "city": "Chennai"
}

file = open("student.json", "w")
json.dump(data, file)
file.close()

print("JSON file created.")
```

```python
import json

file = open("student.json", "r")
data = json.load(file)
file.close()

print(data)
```

# PICKLE FILES

Pickle stores Python objects directly in binary format.

# Terms:

Serialization: Saving objects

Deserialization: Loading objects

```python
import pickle

data = {
    "name": "Zane",
    "age": 4,
    "city": "Chennai"
}

file = open("student.pkl", "wb")
pickle.dump(data, file)
file.close()

print("Pickle file created.")
```

```python
import pickle

file = open("student.pkl", "rb")
data = pickle.load(file)
file.close()

print(data)
```

# CONTEXT MANAGER (with)

The with statement automatically opens and closes the file after use.

```python
# to write a text
with open('sample.txt','w') as file:
    file.write("Hello World")
    file.write("\nHi This is Infant")
    
```

```python
# to read a text
with open('sample.txt', 'r') as file:
    data = file.read()
print(data)
```

# FILE VALIDATION
File validation checks whether

a file exists,

is readable/writable,

is empty,

or has the correct type before using it.

```python
#to check if a file exist

import os

if os.path.exists("sample.txt"):
    print("File is Exists")
else:
    print("File does not Exists")
```

```python
# File Validation

import os

filename = "sample.txt"
if os.path.exists(filename):
    file = open(filename, 'r')
    print(file.read())
    file.close()
else:
    print("File Does not Exists")
```
