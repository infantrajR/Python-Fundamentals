# Module 3 – Exception Handling

## What is Exception Handling?
An exception is an error that happens while a Python program is runnin

Exception handling allows us to handle errors without suddenly stopping the program

Main Keywords are:

try

except

else

finally

raise

assert

What is try?
The try block contains the code that might cause an exception.

# try
What is try?
The try block contains the code that might cause an exception.

```python
try:
    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))

    result = a / b
    print("Result:", result)

except ZeroDivisionError:
    print("Cannot divide by zero.")
```

# except
What is except?

except is used to handle an exception

```python
try:
    number = 10 / 0

except ZeroDivisionError:
    print("Cannot divide by zero")
```

# else
What is else?

The else block executes only when no exception occurs.

```python
try:
    number = 10 / 2
    print(number)

except ZeroDivisionError:
    print("Cannot divide by zero")

else:
    print("Calculation successful")
```

```python
try:
    num = int(input("Enter a number: "))

except ValueError:
    print("Please enter an integer.")

else:
    if num % 2 == 0:
        print("Even Number")
    else:
        print("Odd Number")
```

# finally

What is finally?

finally always executes, whether an exception occurs or not.

```python
try:
    number = 10 / 0

except ZeroDivisionError:
    print("Cannot divide by zero")

finally:
    print("Program completed")
```

# raise
What is raise

raise is used when we want to manually create an exception.

```python
age = 15

if age < 18:
    raise ValueError("Age must be 18 or above")
```

```python
try:
    age = 15

    if age < 18:
        raise ValueError("Age must be 18 or above")

except ValueError as e:
    print("Error:", e)
```

# assert

What is assert?
assert is used to check whether a condition is true.

If the condition is True, the program continues.

If it is False, Python raises AssertionError

```python
age = 20

assert age >= 18

print("Eligible")
```

```python
age = 15

assert age >= 18

print("Eligible")
```

# Custom Exceptions

What is a Custom Exception?

Sometimes our application needs its own error type.

We can create one using a class.

```python
class InvalidAgeError(Exception):
    pass
```

```python
try:
    age = 15

    if age < 18:
        raise InvalidAgeError("Age must be 18 or above")

except InvalidAgeError as e:
    print("Error:", e)
```

# ValueError

When does it occur?

ValueError occurs when a function receives a correct data type but an invalid value.

```python
try:
    age = int("abc")
    print(age)

except ValueError:
    print("Please enter a valid number")
```

# TypeError

When does it occur

TypeError occurs when an operation is performed on an inappropriate data type.

```python
try:
    number = 10
    name = "Vijay"

    result = number + name

except TypeError:
    print("Cannot add integer and string")
```

# IndexError

When does it occur?

IndexError occurs when we try to access an index that does not exist.

```python
numbers = [10, 20, 30]

try:
    print(numbers[5])

except IndexError:
    print("Index does not exist")
```

# KeyError

When does it occur?

KeyError occurs when we try to access a dictionary key that doesn't exist.

```python
student = {
    "name": "Sanjay",
    "age": 25
}

try:
    print(student["salary"])

except KeyError:
    print("Key does not exist")
```

# AttributeError

When does it occur

AttributeError occurs when we try to access an attribute or method that doesn't exist for an object.

```python
name = "Arjun"

try:
    name.append("K")

except AttributeError:
    print("This object does not have append() method")
```

# NameError

When does it occur

NameError occurs when we use a variable or name that has not been defined.

```python
try:
    print(age)

except NameError:
    print("Variable has not been defined")
```

# FileNotFoundError

When does it occur?

It occurs when Python tries to open a file that doesn't exist.

```python
try:
    file = open("student.txt", "r")
    print(file.read())

except FileNotFoundError:
    print("File does not exist")
```

# ZeroDivisionError

When does it occur

It occurs when we try to divide a number by zero

```python
try:
    result = 10 / 0
    print(result)

except ZeroDivisionError:
    print("Cannot divide by zero")
```

# ImportError

When does it occur?

ImportError occurs when Python cannot import something from a module

```python
try:
    from math import abc

except ImportError:
    print("Cannot import abc from math module")
```

# ModuleNotFoundError

When does it occur?

It occurs when Python cannot find the module/package that we are trying to import.

```python
try:
    import my_unknown_module

except ModuleNotFoundError:
    print("Module is not installed or does not exist")
```
