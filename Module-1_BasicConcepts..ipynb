Control Flow

if / elif / else

Step 1 – Understand the Concept

if, elif, and else help Python make decisions.

if checks the first condition.
elif checks another condition if the previous condition is false.
else runs when all the previous conditions are false.
Python checks the conditions from top to bottom.
Once a condition is true, that block is executed and the remaining blocks are skipped.

Example:

Imagine checking a student's marks.

If marks are 80 or above → Grade A
Else if marks are 60 or above → Grade B
Else → Grade C

Python checks each condition one by one and executes the first matching block.

Step 2 – Demonstrate the Concept

```python
marks = 70

if marks >= 80:
    print("Grade A")
elif marks >= 60:
    print("Grade B")
else:
    print("Grade C")
```

Step 3 – Implementation Explanation

What does this program do?

This program takes a student's marks and assigns a grade based on the marks.

80 or above → Grade A
60 or above → Grade B
Below 60 → Grade C
Which part of the code demonstrates the concept?

The following part demonstrates the if / elif / else concept:

if marks >= 80:
    print("Grade A")
elif marks >= 60:
    print("Grade B")
else:
    print("Grade C")
if marks >= 80 checks the first condition.
elif marks >= 60 checks the second condition when the if condition is false.
else handles all remaining cases.
How does the concept work in this implementation?

Python starts by checking the if condition.

If it is true, Python executes the if block and skips the remaining blocks.

If it is false, Python checks the elif condition.

If all the conditions are false, Python executes the else block.

Therefore, only one block is executed in this program.

# List Comprehension

### Step 1 – Understand the Concept

List comprehension is a shorter way to create a new list from an existing sequence.

Normally, we can create a list using a for loop:

```python
numbers = [1, 2, 3, 4, 5] 
squares = [] 
for number in numbers:
    squares.append(number * number)
print(squares)
```

The same thing can be written using list comprehension:

```python
numbers = [1, 2, 3, 4, 5] 
squares = [number * number for number in numbers]
print(squares)
```

# Basic syntax
new_list = [expression for item in sequence]

For example:

squares = [number * number for number in numbers]

Here:

number * number → what we want to store
for number in numbers → goes through each item
[ ] → creates the new list

### Step 2 – Demonstrate the Concept

Example 1 – Create a list of squares

```python
numbers = [1, 2, 3, 4, 5]

squares = [number * number for number in numbers]

print(squares)
```

### Step 3 – Implementation Explanation

The program takes a list of numbers and creates a new list containing only the even numbers.

```python
numbers = [1, 2, 3, 4, 5, 6]

even_numbers = [number for number in numbers if number % 2 == 0]

print(even_numbers)
```

Which part of the code demonstrates the concept?

This is the list comprehension:

[number for number in numbers if number % 2 == 0]

It contains three important parts:

number

The value that will be added to the new list.

for number in numbers

Goes through every number in the original list.

if number % 2 == 0

Allows only even numbers.

How does the concept work in this implementation?

Python goes through the original list one item at a time.

For each number, Python checks:

number % 2 == 0

If the condition is True, that number is added to even_numbers.

So:

1 → odd → skip
2 → even → add
3 → odd → skip
4 → even → add
5 → odd → skip
6 → even → add

The final list becomes:

[2, 4, 6]

# Dictionary Comprehension

### Step 1 – Understand the Concept

Dictionary comprehension is a shorter way to create a new dictionary from an existing sequence or dictionary.

Normally, we can create a dictionary using a for loop:

numbers = [1, 2, 3, 4, 5]

squares = {}

for number in numbers:
    squares[number] = number * number

print(squares)

Output:

{1: 1, 2: 4, 3: 9, 4: 16, 5: 25}

The same thing can be written using dictionary comprehension:

numbers = [1, 2, 3, 4, 5]

squares = {number: number * number for number in numbers}

print(squares)

Output:

{1: 1, 2: 4, 3: 9, 4: 16, 5: 25}

So, dictionary comprehension allows us to create a dictionary in a shorter and cleaner way.

### Real-World Analogy

Imagine you have a list of students:

Ravi
Kumar
Arun

You want to create a dictionary where each student's name is the key and the length of their name is the value:

Ravi  → 4
Kumar → 5
Arun  → 4

Dictionary comprehension can create this dictionary quickly.

### Step 2 – Demonstrate the Concept

```python
numbers = [1, 2, 3, 4, 5] 
squares = {number: number * number for number in numbers} 
print(squares)
```

For example, instead of creating separate functions to add different numbers, we can create one function and give it the numbers we want to add.

The operation is the same, but the input numbers are different.

A Python function works similarly. We can create one function and provide different arguments whenever we call it.

### Step 2 – Demonstrate the Concept

First, let's create a function with parameters:

    total = a + b
    print("Total:", total)

Here, a and b are called parameters.

Now we can provide values when calling the function:

add_numbers(10, 20)

Output:

Total: 30

We can call the same function with different values:

add_numbers(30, 40)
add_numbers(50, 60)

add_numbers(30, 40)

Total: 70
Total: 110
Parameter vs Argument

This is an important distinction.

In the function definition:

def add_numbers(a, b):

a and b are parameters.

When we call:

add_numbers(10, 20)

10 and 20 are arguments.

So:

Parameter → Variable defined in the function

Argument → Actual value passed to the function

## Variable-Length Arguments (*args, **kwargs)
### Definition

Variable-length arguments allow a function to receive an unknown number of inputs.

*args → collects positional arguments into a tuple
**kwargs → collects keyword arguments into a dictionary
Real-Life Example

Imagine a shopping bag.

Sometimes you put 2 items in it, and sometimes you put 10 items.

The bag can handle a different number of items.

```python
def total(*numbers):
    print("Total =", sum(numbers))

total(10, 20, 30, 40)    
```

Example using **kwargs

```python
def employee(**details):
    for key, value in details.items():
        print(key, ":", value)

employee(name="Arun", department="IT", city="Chennai")
```

# Recursive Functions
## Definition

A recursive function is a function that calls itself repeatedly until a condition stops the process.

The stopping condition is called the base case.

Real-Life Example

Imagine opening nested boxes.

Inside one box there is another box, and inside that box another one.

The process continues until you reach the final box.

# Example

```python
def countdown(n):
    if n == 0:
        return
    print(n)
    countdown(n - 1)

countdown(5)
```

# Generators
## Definition

A generator is a special type of function that produces values one at a time instead of creating all the values at once.

Generators use the yield keyword to produce values.

Real-Life Example

Imagine a vending machine.

You select one item, and the machine gives you that item. When you select another item, it gives you the next one.

Similarly, a generator produces one value at a time when needed.

Example

```python
def numbers():
    for i in range(1, 5):
        yield i

gen = numbers()

for value in gen:
    print(value)
```

Here, numbers() is a generator function.

The yield keyword produces one value at a time.

# Yield
## Definition

The yield keyword is used inside a generator function to produce a value and temporarily pause the function.

When the function continues, it resumes from where it stopped.

Real-Life Examplmovie and pressing the pause button.

The movie stops at the current position. When you press play again, it continues from the same position.

Similarly, yield pauses a generator and allows it to continue later.

### Example

```python
def even_numbers():
    for number in range(1, 11):
        if number % 2 == 0:
            yield number

for value in even_numbers():
    print(value)
```

# Decorators
## Definition

Real-Life Example

Similarly, a decorator adds extra behavior to an existing function.

```python
def logger(func):
    def wrapper():
        print("Function started")
        func()
        print("Function completed")

    return wrapper

@logger 
def welcome(): 
    print("Welcome to Python")

welcome()
```

# Closures

```python
def create_multiplier(number): 
    
    def multiply(value):
        return number * value 

    return multiply 

double = create_multiplier(2) 

print(double(8))
```

# Global and nonlocal
## Definition

The global keyword allows a function to modify a variable defined in the global scope.

The nonlocal keyword allows an inner function to modify a variable from its enclosing function.

Real-Life Example

Imagine a company.

global → information shared across the entire company
nonlocal → information shared within a particular department

## Example using global

```python
score = 50 

def update_score(): 
    global score 
    score += 10 
      
update_score() 

print(score)
```

Here, global score allows the function to modify the global variable score.

## Example using nonlocal

```python
def counter(): 
    count = 0 
    
    def increase(): 
        nonlocal count 
        count += 1 
        print(count) 
        
    return increase 
    
result = counter() 
result() 
result() 
result()
```

Here, nonlocal count allows the inner function to modify the count variable from the outer function.

# Type Hinting
### Definition

Type hinting is a way of specifying the expected data types of variables, function parameters, and return values.

### Real-Life Example

Clothes

### Example


```python
def add(a: int, b: int) -> int: 
    return a + b 
print(add(10, 20))
```

b: int → b is expected to be an integer.
-> int → the function is expected to return an integer.
