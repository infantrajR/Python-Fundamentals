Control Flow

if / elif / else

Step 1 – Understand the Concept

if, elif, and else help Python make decisions.

if checks the first condition.
elif checks another condition if the previous condition is false.
else runs when all the previous conditions are false.
Python checks the conditions from top to bottom.
Once a condition is true, that block is executed and the remaining blocks are skipped.

Example:

Imagine checking a student's marks.

If marks are 80 or above → Grade A
Else if marks are 60 or above → Grade B
Else → Grade C

Python checks each condition one by one and executes the first matching block.

Step 2 – Demonstrate the Concept

```python
marks = 70

if marks >= 80:
    print("Grade A")
elif marks >= 60:
    print("Grade B")
else:
    print("Grade C")
```

Step 3 – Implementation Explanation

What does this program do?

This program takes a student's marks and assigns a grade based on the marks.

80 or above → Grade A
60 or above → Grade B
Below 60 → Grade C
Which part of the code demonstrates the concept?

The following part demonstrates the if / elif / else concept:

if marks >= 80:
    print("Grade A")
elif marks >= 60:
    print("Grade B")
else:
    print("Grade C")
if marks >= 80 checks the first condition.
elif marks >= 60 checks the second condition when the if condition is false.
else handles all remaining cases.
How does the concept work in this implementation?

Python starts by checking the if condition.

If it is true, Python executes the if block and skips the remaining blocks.

If it is false, Python checks the elif condition.

If all the conditions are false, Python executes the else block.

Therefore, only one block is executed in this program.

Step 1 – Understand the Concept

A while loop is used to repeat a block of code as long as a condition is true.

while checks a condition before each repetition.
If the condition is True, the code inside the loop runs.
After running the code, Python checks the condition again.
When the condition becomes False, the loop stops.

Example:

Imagine a person climbing stairs.

If there are still stairs left → Climb the next stair.
Continue climbing while stairs are remaining.
When there are no stairs left → Stop.

The while loop works in a similar way. It keeps repeating while its condition is true.

Break Statement

Step 1 – Understand the Concept

The break statement is used to stop a loop immediately.

break stops the current loop.
Python exits the loop when it reaches break.
The loop does not continue with the remaining iterations.
It can be used with both for and while loops.

Example:

Imagine searching for a particular student in a list.

Check the first student.
Check the second student.
When the required student is found → Stop searching.

There is no need to continue checking the remaining students.

The break statement works in the same way. It stops the loop as soon as the required condition is reached.

continue Statement

Step 1 – Understand the Concept

The continue statement is used to skip the current iteration of a loop and move to the next iteration.

continue does not stop the entire loop.
It skips the remaining code for the current iteration.
The loop then moves to the next item.
It can be used with both for and while loops.

Example:

Imagine checking numbers from 1 to 5.

Suppose we don't want to process the number 3.

Check 1 → Process it
Check 2 → Process it
Check 3 → Skip it
Check 4 → Process it
Check 5 → Process it

The continue statement allows us to skip 3 without stopping the entire loop.

Step 2 – Demonstrate the Concept

```python
for number in range(1,10):
    if number ==4:
        continue

    print(number)
```

Step 3 – Implementation Explanation
What does this program do?

This program prints numbers from 1 to 10 but skips the number 4.

Which part of the code demonstrates the concept?

The following code demonstrates continue:

if number == 4:
    continue

When the number is 4, the continue statement skips the rest of that iteration.

How does the concept work in this implementation?

The loop processes the numbers one by one.

For numbers 1,2and 3 the condition is false, so they are printed.

When the number is 4, the condition becomes true.

Python executes continue, so it skips the print() statement for that iteration.

The loop then continues with number 5 and processes the remaining numbers normally.

# List Comprehension

### Step 1 – Understand the Concept

List comprehension is a shorter way to create a new list from an existing sequence.

Normally, we can create a list using a for loop:

```python
numbers = [1, 2, 3, 4, 5] 
squares = [] 
for number in numbers:
    squares.append(number * number)
print(squares)
```

The same thing can be written using list comprehension:

```python
numbers = [1, 2, 3, 4, 5] 
squares = [number * number for number in numbers]
print(squares)
```

# Basic syntax
new_list = [expression for item in sequence]

For example:

squares = [number * number for number in numbers]

Here:

number * number → what we want to store
for number in numbers → goes through each item
[ ] → creates the new list

### Step 2 – Demonstrate the Concept

Example 1 – Create a list of squares

```python
numbers = [1, 2, 3, 4, 5]

squares = [number * number for number in numbers]

print(squares)
```

### Step 3 – Implementation Explanation

The program takes a list of numbers and creates a new list containing only the even numbers.

```python
numbers = [1, 2, 3, 4, 5, 6]

even_numbers = [number for number in numbers if number % 2 == 0]

print(even_numbers)
```

Which part of the code demonstrates the concept?

This is the list comprehension:

[number for number in numbers if number % 2 == 0]

It contains three important parts:

number

The value that will be added to the new list.

for number in numbers

Goes through every number in the original list.

if number % 2 == 0

Allows only even numbers.

How does the concept work in this implementation?

Python goes through the original list one item at a time.

For each number, Python checks:

number % 2 == 0

If the condition is True, that number is added to even_numbers.

So:

1 → odd → skip
2 → even → add
3 → odd → skip
4 → even → add
5 → odd → skip
6 → even → add

The final list becomes:

[2, 4, 6]
