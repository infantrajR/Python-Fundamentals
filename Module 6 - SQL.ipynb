# Database Design
# Database
# Definition

A database is an organized collection of data that allows us to store, manage, and retrieve information efficiently.

# Real-Life Example

Think of a company's employee system.

It may store:

Employee details
Department details
Salary information
Attendance records

All these records can be stored in a database.

# Example
CREATE DATABASE company_db;

This creates a database named company_db.

# Tables
# Definition

A table is a structure inside a database that stores data in rows and columns.

# Real-Life Example

Think of an Excel sheet.

Each column represents a type of information, and each row represents one record.

# Example
CREATE TABLE employees (

    employee_id INT,
    
    name VARCHAR(50),
    
    department VARCHAR(50),
    
    salary INT
);



# Primary Key
# Definition

A primary key is a column or combination of columns that uniquely identifies each row in a table.

A primary key cannot contain duplicate values.

# Real-Life Example

Think of an employee ID.

Every employee has a unique employee ID.

# Example
CREATE TABLE employees (

    employee_id INT PRIMARY KEY,
    
    name VARCHAR(50),
    
    salary INT
);

Here, employee_id uniquely identifies each employee.

# Foreign Key
# Definition

A foreign key is a column that creates a connection between two tables.

It usually refers to the primary key of another table.

Real-Life Example

An employee belongs to a department.

The employee table can store the department ID, which refers to the department table.

# Example
CREATE TABLE departments (
    
    department_id INT PRIMARY KEY,

    department_name VARCHAR(50)

);

CREATE TABLE employees (
    
    employee_id INT PRIMARY KEY,
    
    name VARCHAR(50),
    
    department_id INT,
    
    FOREIGN KEY (department_id)
    
        REFERENCES departments(department_id)
);

Here, department_id in employees is a foreign key.

# Constraints
# Definition

Constraints are rules applied to table columns to control the type of data that can be stored.

Common constraints include:

PRIMARY KEY
FOREIGN KEY
NOT NULL
UNIQUE
DEFAULT
CHECK
Real-Life Example

Think of rules on an online registration form.

For example:

Name cannot be empty.
Email must be unique.
Age must be valid.
# Example
    
CREATE TABLE employees (
    
    employee_id INT PRIMARY KEY,

# Queries

# SELECT
SELECT is used to retrieve data from a table.

Example:

SELECT Name, Age

FROM Students;

This displays the Name and Age columns from the Students table.

# WHERE
WHERE is used to filter rows based on a condition.

# Example:

SELECT *

FROM Students

WHERE Age > 20;

This displays only students whose age is greater than 20.

# ORDER BY
ORDER BY is used to sort the result in ascending (ASC) or descending (DESC) order.

# Example:

SELECT *

FROM Students

ORDER BY Age ASC;

This sorts students by age from smallest to largest.

# GROUP BY
GROUP BY is used to group rows that have the same values in a column.

# Example:

SELECT Department, COUNT(*)

FROM Employees

GROUP BY Department;

This shows the number of employees in each department.

# HAVING
HAVING is used to filter grouped data after using GROUP BY.

# Example:

SELECT Department, COUNT(*)

FROM Employees

GROUP BY Department

HAVING COUNT(*) > 5;

This displays only departments that have more than 5 employees.

# DISTINCT
DISTINCT is used to return only unique values by removing duplicates.

# Example:

SELECT DISTINCT City

FROM Students;

If multiple students are from Chennai, it displays Chennai only once.

# LIMIT
LIMIT is used to restrict the number of rows returned.

# Example:

SELECT *

FROM Students

LIMIT 5;

This displays only the first 5 rows from the Students table.

# INNER JOIN
Returns only the rows that have matching values in both tables that are combined through a common column.

Example:

SELECT Students.Name, Marks.Marks

FROM Students INNER JOIN Marks

ON Students.Student_ID = Marks.Student_ID;

Only matching records from both the tables are retreived

# LEFT JOIN
Returns all rows from the left table and the matching rows from the right table. If there is no match, NULL is returned.

Example:

SELECT Students.Name, Marks.Marks

FROM Students LEFT JOIN Marks

ON Students.Student_ID = Marks.Student_ID;

All the students are shown, even if they have no marks

# RIGHT JOIN
Returns all rows from the right table and the matching rows from the left table. If there is no match, NULL is returned.

Example:

SELECT Students.Name, Marks.Marks

FROM Students RIGHT JOIN Marks

ON Students.Student_ID = Marks.Student_ID;

All records from the Marks table are shown, including Student_ID 104, which has no matching student.

# FULL JOIN
Returns all rows from both tables. If there is no match, NULL is returned for the missing values.

Example:

SELECT Students.Name, Marks.Marks

FROM Students FULL JOIN Marks

ON Students.Student_ID = Marks.Student_ID;

Shows all records from both tables.

# CROSS JOIN
Also called cartesian join, returns every possible combination of rows from both tables.

Example:

SELECT Students.Name, Marks.Marks

FROM Students

CROSS JOIN Marks;

Every student is combined with every marks record. If there are 3 students and 3 marks records, the result contains 9 rows.

# Advanced SQL
# UNION
# Definition

UNION combines the results of two or more SELECT queries and removes duplicate rows.

Real-Life Example

Imagine having employee lists from two different branches and combining them into one list.

Example
SELECT name FROM employees_chennai

UNION
SELECT name FROM employees_bangalore;

The result contains names from both tables without duplicate rows.

#CASE
# Definition

CASE is used to perform conditional logic inside a SQL query.

It works similar to if-elif-else in Python.

# Real-Life Example

Imagine classifying employees based on salary.

Salary >= 60000 → High

Salary >= 40000 → Medium

Otherwise       → Low
# Example
SELECT name, salary,

       CASE
           WHEN salary >= 60000 THEN 'High'
           
           WHEN salary >= 40000 THEN 'Medium'
           
           ELSE 'Low'
           
       END AS salary_level
FROM employees;

# Subqueries
# Definition

A subquery is a query written inside another SQL query.

The inner query provides information used by the outer query.

# Real-Life Example

First find the average salary.

Then find employees whose salary is greater than that average.

# Example
    
SELECT name, salary

FROM employees
    
WHERE salary > (
    
    SELECT AVG(salary)
    
    FROM employees
    
);

The inner query calculates the average salary.

The outer query finds employees earning above that average.

# Common Table Expressions (CTEs)
# Definition

A Common Table Expression, or CTE, is a temporary named result that can be used within a SQL query.

It is created using the WITH keyword.

# Real-Life Example

Imagine preparing an intermediate report first and then using that report to create the final report.

# Example
    
WITH high_salary AS 

    SELECT name, salary

    FROM employees

    WHERE salary > 50000

)

SELECT *

FROM high_salary;

Here, high_salary is the CTE.

CTEs make complex queries easier to read and organize.

# Views
# Definition

A view is a virtual table created from a SQL query.

It does not normally store a separate copy of the underlying data.

Real-Life Example

Imagine creating a saved report that always displays selected information from a larger database.

# Example

CREATE VIEW employee_summary AS

SELECT name, department, salary

FROM employees;

Now we can query the view:

SELECT *
FROM employee_summary;

# Indexes
# Definition

An index is a database structure that helps SQL find rows more efficiently.

Indexes can improve the speed of searches, although they also require storage and can add overhead when data is inserted or updated.

# Real-Life Example

Think of the index at the back of a textbook.

Instead of checking every page to find a topic, you use the index to locate it quickly.

# Example
    
CREATE INDEX idx_employee_name

ON employees(name);


Now the database has an index on the name column.

Indexes are especially useful for columns frequently used in searches, joins, or sorting.

# Window Functions

# Definition

A window function performs a calculation across a group of related rows while keeping the individual rows in the result.

Unlike GROUP BY, window functions do not combine multiple rows into a single row.

# Real-Life Example

Imagine ranking employees by salary.

You want to see:

Each employee

Their salary

Their position compared with other employees

Window functions can provide this information without removing individual employee rows.

# ROW_NUMBER()
# Definition

ROW_NUMBER() assigns a unique sequential number to each row within the specified ordering.

# Example
    
SELECT name, salary,
    
       ROW_NUMBER() OVER (ORDER BY salary DESC) AS row_num

FROM employees;

# Output
name  | salary | row_num
------|--------|--------
Rahul | 60000  | 1
Arun  | 50000  | 2
Priya | 45000  | 3

Each row receives a unique number.

# RANK()
# Definition

RANK() assigns the same rank to rows with equal values.

When there is a tie, the next rank is skipped.

# Example

SELECT name, salary,

       RANK() OVER (ORDER BY salary DESC) AS salary_rank

FROM employees;

For example, if salaries are:

60000

50000

50000

40000

The ranks are:

60000 → 1

50000 → 2

50000 → 2

40000 → 4

# DENSE_RANK()
# Definition

DENSE_RANK() is similar to RANK(), but it does not skip the next rank after a tie.

# Example

For:

60000

50000

50000

40000

The result is:

60000 → 1

50000 → 2

50000 → 2

40000 → 3

Example Query

SELECT name, salary,

       DENSE_RANK() OVER (ORDER BY salary DESC) AS salary_rank

FROM employees;

# LAG()
# Definition

LAG() allows us to access a value from a previous row.

Real-Life Example

Imagine comparing this month's sales with last month's sales.

# Example
    
SELECT month, sales,
    
       LAG(sales) OVER (ORDER BY month) AS previous_sales

FROM monthly_sales;

Example result:

month | sales | previous_sales
------|-------|---------------
Jan   | 50000 | NULL
Feb   | 55000 | 50000
Mar   | 60000 | 55000

For January, there is no previous month, so the result is NULL.

# LEAD()
# Definition

LEAD() allows us to access a value from the next row.

Real-Life Example

Imagine comparing this month's sales with next month's sales.

# Example
    
SELECT month, sales,
    
       LEAD(sales) OVER (ORDER BY month) AS next_sales

FROM monthly_sales;

Example result:

month | sales | next_sales
------|-------|-----------
Jan   | 50000 | 55000
Feb   | 55000 | 60000
Mar   | 60000 | NULL

For March, there is no next month, so the result is NULL.
